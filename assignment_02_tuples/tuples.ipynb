```python
## Q11. Tuple Indexing
colors = ("red", "green", "blue", "yellow", "purple")

second = colors[1]
last = colors[-1]
mid = len(colors) // 2
middle_three = colors[mid - 1 : mid + 2]

print(second)
print(last)
print(middle_three)
```

```python
# Q12. Tuple Unpacking
employee = ("Ali", 28, "Data Engineer", 150000)

name, age, job, salary = employee

print("Name:", name)
print("Salary:", salary)
```

```python
# Q13. Tuple Concatenation
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

combined = tuple1 + tuple2
reversed_tuple = combined[::-1]

print(combined)
print(reversed_tuple)
```

```python
# Q14. Count and Locate
numbers = (10, 20, 30, 20, 40, 20, 50)

count_20 = numbers.count(20)
first_index = numbers.index(20)

print(count_20)
print(first_index)
```

```python
# Q15. Extended Tuple Unpacking
data = (100, 200, 300, 400, 500, 600)

first, second, *remaining, last = data

print(first, second, last)
print(remaining)
```

```python
# Q16. Modify an Immutable Tuple
numbers = (10, 20, 30, 40, 50)

i = numbers.index(30)
new_numbers = numbers[:i] + (300,) + numbers[i + 1:]

print(new_numbers)
print(numbers)
```

```python
# Q17. Nested Tuple Access
students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)

sara_second = students[1][1][1]
ali_average = sum(students[0][1]) / len(students[0][1])
ahmed_highest = max(students[2][1])

print(sara_second)
print(ali_average)
print(ahmed_highest)
```

```python
# Q18. Tuple Sorting
from operator import itemgetter

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

sorted_employees = tuple(sorted(employees, key=itemgetter(1), reverse=True))

print(sorted_employees)
```

```python
# Q19. Tuple Matrix
import math

matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])
diagonal_product = math.prod(diagonal)
row_sums = (sum(matrix[0]), sum(matrix[1]), sum(matrix[2]))

print(diagonal)
print(diagonal_product)
print(row_sums)
```

```python
# Q20. Inventory Tuple Challenge
from operator import itemgetter

records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

total_value = (
    records[0][1] * records[0][2]
    + records[1][1] * records[1][2]
    + records[2][1] * records[2][2]
)

most_expensive = max(records, key=itemgetter(1))[0]

names = (records[0][0], records[1][0], records[2][0])
reversed_names = names[::-1]

print(total_value)
print(most_expensive)
print(reversed_names)
```
