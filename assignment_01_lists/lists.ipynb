## Python Lists

```python
# Q1. Accessing list elements
numbers = [10, 20, 30, 40, 50, 60, 70]
first = numbers[0]
middle = numbers[len(numbers) // 2]  
last = numbers[-1]
print(first, middle, last)       
```

```python
# Q2. List slicing
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
first_five = numbers[:5]
last_four = numbers[-4:]
every_second_reverse = numbers[::-2]  
print(first_five)           
print(last_four)           
print(every_second_reverse)
```

```python
# Q3. List modification
fruits = ["apple", "banana", "cherry", "mango"]
fruits[fruits.index("banana")] = "orange"  
fruits.insert(2, "grapes")                  
fruits.remove("cherry")                     
print(fruits)  
```

```python
# Q4. Reverse without reverse()
numbers = [12, 45, 23, 67, 89, 34]
reversed_numbers = numbers[::-1]   
print(reversed_numbers)  
print(numbers)           
```

```python
# Q5. Second largest unique number
numbers = [45, 12, 89, 34, 89, 67, 45, 23]
unique_sorted = sorted(set(numbers))   
second_largest = unique_sorted[-2]
print(second_largest)   
```

```python
# Q6. Rotate a list
numbers = [10, 20, 30, 40, 50, 60]
k = 2
k = k % len(numbers)    
rotated = numbers[-k:] + numbers[:-k]
print(rotated)   
```

```python
# Q7. Remove duplicates, keep order
numbers = [10, 20, 10, 30, 20, 40, 30, 50]
no_duplicates = list(dict.fromkeys(numbers))   # dict keeps first-seen order
print(no_duplicates) 
```

```python
# Q8. Flatten a nested list
data = [[10, 20], [30, 40], [50, 60]]
flat = sum(data, [])    # adds the inner lists together starting from an empty list
print(flat)    
```

```python
# Q9. Extract data from a matrix
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

main_diagonal = [matrix[0][0], matrix[1][1], matrix[2][2]]
secondary_diagonal = [matrix[0][2], matrix[1][1], matrix[2][0]]
middle_column = [matrix[0][1], matrix[1][1], matrix[2][1]]

print(main_diagonal, sum(main_diagonal))           
print(secondary_diagonal, sum(secondary_diagonal)) 
print(middle_column, sum(middle_column)) 
```

```python
# Q10. List transformation challenge
data = [15, 5, 25, 10, 35, 20, 30]
unique_sorted = sorted(set(data))        
three_largest = unique_sorted[-3:]    
two_smallest_desc = unique_sorted[:2][::-1]  

result = three_largest + two_smallest_desc
print(result)  
```
