```python
# Q41. Student Performance System
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

averages = {
    students[0]["name"]: sum(students[0]["marks"]) / len(students[0]["marks"]),
    students[1]["name"]: sum(students[1]["marks"]) / len(students[1]["marks"]),
    students[2]["name"]: sum(students[2]["marks"]) / len(students[2]["marks"])
}

top_student = max(averages, key=averages.get)
names = (students[0]["name"], students[1]["name"], students[2]["name"])
unique_marks = set(students[0]["marks"] + students[1]["marks"] + students[2]["marks"])

print(averages)
print(top_student)
print(names)
print(unique_marks)
```

```python
# Q42. Employee Skills
employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

ali = employees["Ali"]
sara = employees["Sara"]
ahmed = employees["Ahmed"]

common_skills = ali & sara & ahmed
ali_only = ali - (sara | ahmed)
all_skills = ali | sara | ahmed

required = {"Python", "SQL"}
meets_requirement = {
    "Ali": required.issubset(ali),
    "Sara": required.issubset(sara),
    "Ahmed": required.issubset(ahmed)
}

print(common_skills)
print(ali_only)
print(all_skills)
print(meets_requirement)
```

```python
# Q46. E-Commerce Transactions
transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

items1 = transactions[0]["items"]
items2 = transactions[1]["items"]
items3 = transactions[2]["items"]

total1 = items1[0][1] * items1[0][2] + items1[1][1] * items1[1][2]
total2 = items2[0][1] * items2[0][2] + items2[1][1] * items2[1][2]
total3 = items3[0][1] * items3[0][2]

total_revenue = total1 + total2 + total3
ali_spending = total1 + total3

transaction_totals = {
    transactions[0]["id"]: total1,
    transactions[1]["id"]: total2,
    transactions[2]["id"]: total3
}
largest_id = max(transaction_totals, key=transaction_totals.get)
largest_amount = transaction_totals[largest_id]

unique_products = {
    items1[0][0], items1[1][0],
    items2[0][0], items2[1][0],
    items3[0][0]
}

print(total_revenue)
print(ali_spending)
print(largest_id, largest_amount)
print(unique_products)
```

```python
# Q47. Pipeline Monitoring
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

pipeline_totals = {
    "pipeline_1": sum(pipelines["pipeline_1"]["records"]),
    "pipeline_2": sum(pipelines["pipeline_2"]["records"]),
    "pipeline_3": sum(pipelines["pipeline_3"]["records"])
}

total_processed = sum(pipeline_totals.values())

failed = []
if pipelines["pipeline_1"]["status"] == "failed":
    failed.append("pipeline_1")
if pipelines["pipeline_2"]["status"] == "failed":
    failed.append("pipeline_2")
if pipelines["pipeline_3"]["status"] == "failed":
    failed.append("pipeline_3")

largest_pipeline = max(pipeline_totals, key=pipeline_totals.get)

average_per_run = {
    "pipeline_1": pipeline_totals["pipeline_1"] / len(pipelines["pipeline_1"]["records"]),
    "pipeline_2": pipeline_totals["pipeline_2"] / len(pipelines["pipeline_2"]["records"]),
    "pipeline_3": pipeline_totals["pipeline_3"] / len(pipelines["pipeline_3"]["records"])
}

print(total_processed)
print(failed)
print(largest_pipeline, pipeline_totals[largest_pipeline])
print(average_per_run)
```

```python
# Q48. Banking Transactions
from bisect import bisect_left

accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

balances = {
    "A101": sum(accounts["A101"]["transactions"]),
    "A102": sum(accounts["A102"]["transactions"]),
    "A103": sum(accounts["A103"]["transactions"])
}

highest_id = max(balances, key=balances.get)
highest_holder = accounts[highest_id]["holder"]


def split_transactions(tx):
    s = sorted(tx)
    i = bisect_left(s, 0)
    return s[:i], s[i:]


w1, d1 = split_transactions(accounts["A101"]["transactions"])
w2, d2 = split_transactions(accounts["A102"]["transactions"])
w3, d3 = split_transactions(accounts["A103"]["transactions"])

total_withdrawals = -(sum(w1) + sum(w2) + sum(w3))
largest_deposit = max(max(d1), max(d2), max(d3))

print(balances)
print(highest_holder, balances[highest_id])
print(total_withdrawals)
print(largest_deposit)
```

```python
# Q49. Product Recommendations
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

ali_p = users["Ali"]["purchased"]
sara_p = users["Sara"]["purchased"]
ahmed_p = users["Ahmed"]["purchased"]

ali_w = users["Ali"]["wishlist"]
sara_w = users["Sara"]["wishlist"]
ahmed_w = users["Ahmed"]["wishlist"]

bought_by_two = (ali_p & sara_p) | (ali_p & ahmed_p) | (sara_p & ahmed_p)
ali_wishlist_left = ali_w - ali_p
shared_wishlist = (ali_w & sara_w) | (ali_w & ahmed_w) | (sara_w & ahmed_w)
all_products = ali_p | sara_p | ahmed_p | ali_w | sara_w | ahmed_w

print(bought_by_two)
print(ali_wishlist_left)
print(shared_wishlist)
print(all_products)
```

```python
# Q50. Final Company Analysis
from operator import itemgetter

company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

eng = company["departments"]["Engineering"]
ana = company["departments"]["Analytics"]
eng_emps = eng["employees"]
ana_emps = ana["employees"]
all_emps = eng_emps + ana_emps

# 1. Payroll per department
payroll = {
    "Engineering": eng_emps[0][1] + eng_emps[1][1],
    "Analytics": ana_emps[0][1] + ana_emps[1][1]
}
print(payroll)

# 2. Company payroll
company_payroll = sum(payroll.values())
print(company_payroll)

# 3. Highest-paid employee
highest_paid = max(all_emps, key=itemgetter(1))
print(highest_paid[0], highest_paid[1])

# 4. Skills shared by all
shared_skills = all_emps[0][2] & all_emps[1][2] & all_emps[2][2] & all_emps[3][2]
print(shared_skills)

# 5. Engineering-only skills
eng_skills = eng_emps[0][2] | eng_emps[1][2]
ana_skills = ana_emps[0][2] | ana_emps[1][2]
eng_only_skills = eng_skills - ana_skills
print(eng_only_skills)

# 6. Remaining budget per department
remaining = {
    "Engineering": eng["budget"] - payroll["Engineering"],
    "Analytics": ana["budget"] - payroll["Analytics"]
}
print(remaining)

# 7. Department with higher average salary
avg_salary = {
    "Engineering": payroll["Engineering"] / len(eng_emps),
    "Analytics": payroll["Analytics"] / len(ana_emps)
}
higher_avg_dept = max(avg_salary, key=avg_salary.get)
print(higher_avg_dept)

# 8. Name-to-salary dictionary
name_to_salary = {
    all_emps[0][0]: all_emps[0][1],
    all_emps[1][0]: all_emps[1][1],
    all_emps[2][0]: all_emps[2][1],
    all_emps[3][0]: all_emps[3][1]
}
print(name_to_salary)

# 9. All unique skills
all_unique_skills = eng_skills | ana_skills
print(all_unique_skills)

# 10. Is combined budget sufficient?
total_budget = eng["budget"] + ana["budget"]
is_sufficient = total_budget >= company_payroll
print(is_sufficient)
```
