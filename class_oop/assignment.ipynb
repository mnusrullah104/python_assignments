## Q1. Student Result Manager

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self._marks = 0
        self.__grade = "F"
        self.set_marks(marks)

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self._marks = marks
            if marks >= 80:
                self.__grade = "A"
            elif marks >= 60:
                self.__grade = "B"
            elif marks >= 40:
                self.__grade = "C"
            else:
                self.__grade = "F"
            print("Marks updated to", marks)
        else:
            print("Invalid marks! Marks must be between 0 and 100")

    def get_marks(self):
        return self._marks

    def get_grade(self):
        return self.__grade


s1 = Student("Ali", 85)
print(s1.name, s1.get_marks(), s1.get_grade())

s1.set_marks(65)
print(s1.name, s1.get_marks(), s1.get_grade())

s1.set_marks(120)
print(s1.name, s1.get_marks(), s1.get_grade())

s2 = Student("Sara", 35)
print(s2.name, s2.get_marks(), s2.get_grade())

try:
    print(s1.__grade)
except AttributeError as e:
    print("Error:", e)
    print("Python renames __grade to _Student__grade (name mangling), so direct access fails")
    print("Still reachable as:", s1._Student__grade)
```

## Q2. Secure Bank Account

```python
class BankAccount:
    def __init__(self, holder_name, balance, pin):
        self.holder_name = holder_name
        self.__balance = balance
        self.__pin = pin

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit rejected: amount must be greater than 0")
            return
        self.__balance += amount
        print("Deposited", amount, "| New balance:", self.__balance)

    def withdraw(self, amount, pin):
        if pin != self.__pin:
            print("Withdrawal rejected: incorrect PIN")
        elif amount <= 0:
            print("Withdrawal rejected: amount must be greater than 0")
        elif amount > self.__balance:
            print("Withdrawal rejected: insufficient balance")
        else:
            self.__balance -= amount
            print("Withdrew", amount, "| New balance:", self.__balance)

    def get_balance(self, pin):
        if pin == self.__pin:
            return self.__balance
        print("Incorrect PIN")
        return None


acc = BankAccount("Ahmed", 5000, 1234)

acc.deposit(2000)
acc.withdraw(1500, 1234)

acc.deposit(-500)
acc.withdraw(1000, 9999)
acc.withdraw(50000, 1234)

print("Balance:", acc.get_balance(1234))
acc.get_balance(1111)

print("Balance read without PIN:", acc._BankAccount__balance)
print("PIN read without permission:", acc._BankAccount__pin)
print("Double underscore only renames the attribute (name mangling), it does not encrypt or truly hide it")
```

## Q3. Employee Roles

```python
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def display_details(self):
        print("Name:", self.name)
        print("Employee ID:", self.emp_id)


class Developer(Employee):
    def __init__(self, name, emp_id, language):
        super().__init__(name, emp_id)
        self.language = language

    def display_details(self):
        super().display_details()
        print("Role: Developer")
        print("Programming Language:", self.language)


class Manager(Employee):
    def __init__(self, name, emp_id, team_size):
        super().__init__(name, emp_id)
        self.team_size = team_size

    def display_details(self):
        super().display_details()
        print("Role: Manager")
        print("Team Size:", self.team_size)


dev = Developer("Usman", 101, "Python")
mgr = Manager("Hina", 102, 8)

dev.display_details()
print("-" * 25)
mgr.display_details()
```

## Q4. Vehicle Rental System

```python
class Vehicle:
    def __init__(self, brand, model, daily_rate):
        self.brand = brand
        self.model = model
        self.daily_rate = daily_rate

    def calculate_rent(self, days):
        if days <= 0:
            raise ValueError("Rental days must be greater than 0")
        return self.daily_rate * days


class Car(Vehicle):
    def calculate_rent(self, days):
        base = super().calculate_rent(days)
        return base + 1000


class Bike(Vehicle):
    def calculate_rent(self, days):
        base = super().calculate_rent(days)
        return base - base * 0.10


car = Car("Toyota", "Corolla", 5000)
bike = Bike("Honda", "CD 70", 1000)

print(car.brand, car.model, "- 3 days rent: Rs.", car.calculate_rent(3))
print(bike.brand, bike.model, "- 3 days rent: Rs.", bike.calculate_rent(3))

try:
    car.calculate_rent(0)
except ValueError as e:
    print("Error:", e)

try:
    bike.calculate_rent(-2)
except ValueError as e:
    print("Error:", e)
```

## Q5. Notification Dispatcher

```python
class EmailNotification:
    def send(self, message):
        print("Sending EMAIL:", message)


class SMSNotification:
    def send(self, message):
        print("Sending SMS:", message)


class PushNotification:
    def send(self, message):
        print("Sending PUSH notification:", message)


def notify_all(notifications, message):
    for n in notifications:
        n.send(message)


notify_all([EmailNotification(), SMSNotification(), PushNotification()], "Your order has been shipped")

print("-" * 25)


class WhatsAppNotification:
    def send(self, message):
        print("Sending WHATSAPP message:", message)


notify_all([EmailNotification(), WhatsAppNotification()], "Meeting at 5 PM")

print("Polymorphism: notify_all() calls n.send() on every object, and each class runs its own version of send()")
print("WhatsAppNotification worked without changing notify_all()")
```

## Q6. Checkout Discount Calculator

```python
class RegularCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative")
        return amount - amount * 0.05


class PremiumCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative")
        return amount - amount * 0.10


class VIPCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative")
        return amount - amount * 0.20


def print_receipt(customer, amount):
    try:
        final = customer.final_price(amount)
        print(type(customer).__name__, "| Bill: Rs.", amount, "| Final: Rs.", final)
    except ValueError as e:
        print("Error:", e)


for c in [RegularCustomer(), PremiumCustomer(), VIPCustomer()]:
    print_receipt(c, 10000)

print_receipt(RegularCustomer(), -500)

print("-" * 25)


class GoldCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative")
        return amount - amount * 0.15


print_receipt(GoldCustomer(), 10000)
```

## Q7. Shape Calculator

```python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

    def perimeter(self):
        return 2 * 3.14159 * self.radius


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


circle = Circle(5)
print("Circle  -> Area:", circle.area(), "| Perimeter:", circle.perimeter())

rect = Rectangle(4, 6)
print("Rectangle -> Area:", rect.area(), "| Perimeter:", rect.perimeter())

try:
    s = Shape()
except TypeError as e:
    print("Error:", e)
    print("Shape is abstract and has unimplemented abstract methods, so Python does not allow creating its object")
```

## Q8. Payment Gateway

```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass


class CardPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        print("Method: Card | Type: Payment | Amount: Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        print("Method: Card | Type: Refund | Amount: Rs.", amount)


class WalletPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        print("Method: Wallet | Type: Payment | Amount: Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        print("Method: Wallet | Type: Refund | Amount: Rs.", amount)


class BankTransferPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        print("Method: Bank Transfer | Type: Payment | Amount: Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        print("Method: Bank Transfer | Type: Refund | Amount: Rs.", amount)


def process_transaction(method, amount):
    try:
        method.pay(amount)
    except ValueError as e:
        print("Transaction failed:", e)


for m in [CardPayment(), WalletPayment(), BankTransferPayment()]:
    process_transaction(m, 2500)

process_transaction(CardPayment(), -100)
CardPayment().refund(500)

print("-" * 25)


class CryptoPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        print("Method: Crypto | Type: Payment | Amount: Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            raise ValueError("Amount must be positive")
        print("Method: Crypto | Type: Refund | Amount: Rs.", amount)


process_transaction(CryptoPayment(), 3000)
```
