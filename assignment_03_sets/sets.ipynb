```python
# Q21. Unique Values
numbers = [10, 20, 10, 30, 40, 20, 50]

unique_numbers = set(numbers)
count_unique = len(unique_numbers)

print(unique_numbers)
print(count_unique)
```

```python
# Q22. Set Operations
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

union = a | b
intersection = a & b
difference = a - b
sym_difference = a ^ b

print(union)
print(intersection)
print(difference)
print(sym_difference)
```

```python
# Q23. Add and Remove
skills = {"Python", "SQL", "AWS"}

skills.add("Snowflake")
skills.add("Databricks")
skills.discard("AWS")

print(skills)
```

```python
# Q24. Common Students
python_students = {"Ali", "Sara", "Ahmed", "Zain"}
sql_students = {"Sara", "Ahmed", "Hamza", "Usman"}

both = python_students & sql_students
exactly_one = python_students ^ sql_students

print(both)
print(exactly_one)
```

```python
# Q25. Subset and Superset
required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}

is_subset = required_skills.issubset(candidate_skills)
is_superset = candidate_skills.issuperset(required_skills)

print(is_subset)
print(is_superset)
```

```python
# Q26. Missing Numbers
numbers = {1, 2, 4, 6, 8, 10}

all_numbers = set(range(1, 11))
missing = all_numbers - numbers

print(missing)
```

```python
# Q27. Three-Set Analysis
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

all_three = python & sql & aws
python_sql_only = (python & sql) - aws
at_least_two = (python & sql) | (python & aws) | (sql & aws)

print(all_three)
print(python_sql_only)
print(at_least_two)
```

```python
# Q28. Transaction Analysis
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

all_days = day1 & day2 & day3
at_least_two = (day1 & day2) | (day1 & day3) | (day2 & day3)
exactly_one = (day1 | day2 | day3) - at_least_two

print(all_days)
print(exactly_one)
print(at_least_two)
```

```python
# Q29. Access Control
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing = required_permissions - user_permissions
extra = user_permissions - required_permissions
fully_authorized = required_permissions.issubset(user_permissions)

print(missing)
print(extra)
print(fully_authorized)
```

```python
# Q30. Customer Segmentation
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website & mobile & store
at_least_two = (website & mobile) | (website & store) | (mobile & store)
exactly_two = at_least_two - all_three
exactly_one = (website | mobile | store) - at_least_two

print(exactly_one)
print(exactly_two)
print(all_three)
```
