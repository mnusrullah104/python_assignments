```python
# Q31. Dictionary Access
student = {"name": "Ali", "age": 22, "course": "Python", "marks": 85}

name = student["name"]
marks = student["marks"]
city = student.get("city", "Karachi")

print(name)
print(marks)
print(city)
```

```python
# Q32. Update Dictionary
employee = {"name": "Sara", "salary": 100000, "department": "IT"}

employee["salary"] = employee["salary"] * 1.2
employee["bonus"] = 15000
employee["department"] = "Data Engineering"

print(employee)
```

```python
# Q33. Dictionary Methods
product = {"id": 101, "name": "Laptop", "price": 120000, "stock": 5}

keys = product.keys()
values = product.values()
items = product.items()
removed_stock = product.pop("stock")

print(keys)
print(values)
print(items)
print(removed_stock)
print(product)
```

```python
# Q34. Merge Dictionaries
dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 50, "d": 40}

merged = dict1 | dict2

print(merged)
```

```python
# Q35. Student Marks
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}

highest = max(marks.values())
lowest = min(marks.values())
average = sum(marks.values()) / len(marks)
top_student = max(marks, key=marks.get)

print(highest)
print(lowest)
print(average)
print(top_student)
```

```python
# Q36. Nested Employee Records
employees = {
    "E101": {"name": "Ali", "salary": 80000},
    "E102": {"name": "Sara", "salary": 120000},
    "E103": {"name": "Ahmed", "salary": 95000}
}

sara_salary = employees["E102"]["salary"]
employees["E101"]["salary"] = employees["E101"]["salary"] * 1.15
total_payroll = (
    employees["E101"]["salary"]
    + employees["E102"]["salary"]
    + employees["E103"]["salary"]
)

print(sara_salary)
print(employees["E101"]["salary"])
print(total_payroll)
```

```python
# Q37. Dictionary Inversion
data = {"Python": "Ali", "SQL": "Sara", "AWS": "Ahmed"}

inverse = dict(zip(data.values(), data.keys()))

print(inverse)
```

```python
# Q38. Order Calculation
order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}

laptop_subtotal = order["laptop"]["price"] * order["laptop"]["quantity"]
mouse_subtotal = order["mouse"]["price"] * order["mouse"]["quantity"]
keyboard_subtotal = order["keyboard"]["price"] * order["keyboard"]["quantity"]

grand_total = laptop_subtotal + mouse_subtotal + keyboard_subtotal
after_discount = grand_total * 0.90

print(laptop_subtotal, mouse_subtotal, keyboard_subtotal)
print(grand_total)
print(after_discount)
```

```python
# Q39. Employee Performance
employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

averages = {
    "Ali": sum(employees["Ali"].values()) / len(employees["Ali"]),
    "Sara": sum(employees["Sara"].values()) / len(employees["Sara"]),
    "Ahmed": sum(employees["Ahmed"].values()) / len(employees["Ahmed"])
}

top_performer = max(averages, key=averages.get)
highest_sql = max(
    employees["Ali"]["sql"],
    employees["Sara"]["sql"],
    employees["Ahmed"]["sql"]
)

print(averages)
print(top_performer)
print(highest_sql)
```

```python
# Q40. Inventory Management
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

item_values = {
    "P101": inventory["P101"]["price"] * inventory["P101"]["stock"],
    "P102": inventory["P102"]["price"] * inventory["P102"]["stock"],
    "P103": inventory["P103"]["price"] * inventory["P103"]["stock"]
}

total_value = sum(item_values.values())
most_valuable_id = max(item_values, key=item_values.get)
most_valuable_item = inventory[most_valuable_id]["name"]

inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5

print(total_value)
print(most_valuable_item)
print(inventory)
```
